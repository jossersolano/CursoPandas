### **1. Métodos para Resumen y Estadísticas**
- **`count()`**: Cuenta el número de elementos no nulos en la Serie.
- **`sum()`**: Suma los valores.
- **`mean()`**: Calcula el promedio.
- **`median()`**: Calcula la mediana.
- **`min()` / `max()`**: Encuentra el valor mínimo o máximo.
- **`std()`**: Devuelve la desviación estándar.
- **`describe()`**: Proporciona estadísticas descriptivas (promedio, desviación estándar, etc.).
- **`value_counts()`**: Cuenta las ocurrencias de cada valor único.


### **2. Métodos para Transformaciones**
- **`apply(func)`**: Aplica una función personalizada a los elementos de la Serie.
- **`map(func)`**: Mapea una función (o diccionario) a los valores.
- **`astype(dtype)`**: Convierte los valores a un tipo específico (e.g., `int`, `float`, `str`).
- **`round(decimals)`**: Redondea los valores.
- **`clip(lower, upper)`**: Limita los valores a un rango.
- **`replace(to_replace, value)`**: Reemplaza valores específicos.


### **3. Métodos para Manipulación de Datos**
- **`dropna()`**: Elimina valores nulos (`NaN`) de la Serie.
- **`fillna(value)`**: Rellena valores nulos con un valor dado.
- **`isnull()` / `notnull()`**: Verifica si los elementos son nulos (`True` o `False`).
- **`sort_values()`**: Ordena los valores.
- **`unique()`**: Obtiene los valores únicos.
- **`duplicated()`**: Identifica valores duplicados.


### **4. Métodos para Índices**
- **`reset_index()`**: Resetea el índice de la Serie.
- **`set_axis(labels)`**: Cambia los índices.
- **`rename(index=new_name)`**: Renombra los índices.


### **5. Métodos para Texto (si los datos son cadenas)**
- **`str.contains()`**: Revisa si las cadenas contienen un patrón.
- **`str.lower()` / `str.upper()`**: Convierte texto a minúsculas/mayúsculas.
- **`str.strip()`**: Elimina espacios al inicio y al final.
- **`str.replace(pat, repl)`**: Reemplaza patrones por otro valor.
- **`str.split()`**: Divide las cadenas.


### **6. Métodos para Trabajo con Tiempos (si la Serie contiene datos de fecha)**
- **`dt.year`, `dt.month`, `dt.day`**: Extrae año, mes, día, etc.
- **`dt.strftime(format)`**: Convierte fechas a cadenas con un formato específico.
- **`dt.weekday`**: Obtiene el día de la semana como un número.
