### Para limpiar datos usar:
1. Filas con valores vacíos (NaN o None)
```python
# Todas las filas que tengan al menos un NaN/None
df[df.isnull().any(axis=1)]

# O solo en una columna específica (ejemplo: fecha_retiro)
df[df["fecha_retiro"].isnull()]
```

2. Fechas incorrectas
```python
# Intentar convertir a datetime y marcar errores como NaT
df["fecha_ingreso"] = pd.to_datetime(df["fecha_ingreso"], errors="coerce")

# Filas con fechas inválidas (NaT)
df[df["fecha_ingreso"].isna()]
```

3. Números duplicados
```python
# Filas duplicadas en codigo_contrato
df[df["codigo_contrato"].duplicated(keep=False)]

# o ver todo el df con filas duplicadas
df[df.duplicated(keep=False)]
```

4. para limpiar
```python
df = df.dropna()              # elimina filas con NaN
df = df.drop_duplicates()     # elimina filas duplicadas

# reemplazar manualmente valores incorrectos por nulos en fechas
df.loc[df["fecha_ingreso"] < "2020-01-01", "fecha_ingreso"] = pd.NaT
```
