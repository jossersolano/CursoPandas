```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

# Número de filas
n = 50000

# IDs
ids = np.arange(1, n+1)

# Nombres ficticios
nombres = [f"Trabajador_{i}" for i in ids]

# Fechas de ingreso (2020-2025 con crecimiento progresivo)
years = np.random.choice([2020, 2021, 2022, 2023, 2024, 2025],
                         size=n,
                         p=[0.05, 0.1, 0.15, 0.2, 0.25, 0.25])  # más ingresos en años recientes
fechas_ingreso = [datetime(year, random.randint(1,12), random.randint(1,28)) for year in years]

# Edad
edades = np.random.randint(18, 66, size=n)

# Sexo
sexos = np.random.choice(["M", "F"], size=n)

# Fecha de retiro (90% con fecha, 10% activos)
fecha_retiro = []
for fi in fechas_ingreso:
    if random.random() < 0.9:  # retirado
        fr = fi + timedelta(days=random.randint(30, 1500))
        if fr > datetime(2025,12,31):
            fr = None
        fecha_retiro.append(fr)
    else:  # activo
        fecha_retiro.append(None)

# Cargo
cargos = np.random.choice(
    ["Analista","Ingeniero","Gerente","Auxiliar","Director","Coordinador","Consultor"],
    size=n
)

# Construir DataFrame
df = pd.DataFrame({
    "id": ids,
    "nombre": nombres,
    "fecha_ingreso": fechas_ingreso,
    "edad": edades,
    "sexo": sexos,
    "fecha_retiro": fecha_retiro,
    "cargo": cargos
})

# Exportar a CSV
df.to_csv("empleados_sinteticos.csv", index=False)
print("Archivo generado: empleados_sinteticos.csv")
```
